```python
# ============================================================
# MLP EQUALIZER FOR 16-QAM / OFDM / UAV CHANNEL
#
# Complete chain:
# Bits
#   ↓
# 16-QAM Mapper
#   ↓
# OFDM Modulator
#   ↓
# Multipath + Doppler Channel
#   ↓
# Carrier Frequency Offset
#   ↓
# PA Nonlinearity
#   ↓
# Mixer IQ Imbalance
#   ↓
# LNA Gain + Noise + Nonlinearity
#   ↓
# AWGN + Colored Noise
#   ↓
# MLP Equalizer
#   ↓
# 16-QAM Demodulator
#   ↓
# BER / EVM / SER
#
# Python 3.x
# ============================================================

import numpy as np
import matplotlib.pyplot as plt

from sklearn.neural_network import MLPRegressor
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.metrics import mean_squared_error

from scipy.signal import lfilter


# ============================================================
# 1. GLOBAL PARAMETERS
# ============================================================

SEED = 42
rng = np.random.default_rng(SEED)

# Modulation
M = 16
BITS_PER_SYMBOL = int(np.log2(M))

# OFDM
N_SUBCARRIERS = 64
CP_LEN = 16

# Training / testing
N_TRAIN_OFDM = 250
N_TEST_OFDM = 50

# MLP equalizer
EQUALIZER_MEMORY = 3

# SNR range
TRAIN_SNR_DB = (8, 25)
TEST_SNR_DB = 15

# Channel
NUM_PATHS = 6
MAX_DELAY_SAMPLES = 8

# Doppler
DOPPLER_MIN_HZ = 0
DOPPLER_MAX_HZ = 300

# CFO
CFO_MIN = -0.02
CFO_MAX = 0.02

# Colored noise
COLORED_NOISE_STRENGTH = 0.35

# Hardware impairments
PA_NONLINEARITY = 0.15
LNA_NONLINEARITY = 0.05
IQ_GAIN_IMBALANCE_DB = 0.7
IQ_PHASE_IMBALANCE_DEG = 3.0

# Sampling frequency
FS = 15.36e6


# ============================================================
# 2. 16-QAM CONSTELLATION
# ============================================================

def qam16_constellation():
    levels = np.array([-3, -1, 1, 3])
    constellation = []

    for i in levels:
        for q in levels:
            constellation.append(i + 1j * q)

    constellation = np.array(constellation)

    # Normalize average symbol power to 1
    constellation /= np.sqrt(np.mean(np.abs(constellation) ** 2))

    return constellation


CONSTELLATION = qam16_constellation()


# ============================================================
# 3. GRAY-CODED 16-QAM MAPPING
# ============================================================

def gray_to_level(two_bits):
    mapping = {
        (0, 0): -3,
        (0, 1): -1,
        (1, 1): 1,
        (1, 0): 3
    }

    return mapping[tuple(two_bits)]


def bits_to_16qam(bits):

    bits = np.asarray(bits).reshape(-1, 4)

    symbols = []

    for b in bits:

        i = gray_to_level(b[:2])
        q = gray_to_level(b[2:])

        symbols.append(i + 1j * q)

    symbols = np.asarray(symbols)

    symbols /= np.sqrt(10)

    return symbols


# ============================================================
# 4. 16-QAM DEMODULATION
# ============================================================

def level_to_gray(value):

    if value == -3:
        return [0, 0]

    if value == -1:
        return [0, 1]

    if value == 1:
        return [1, 1]

    return [1, 0]


def demodulate_16qam(symbols):

    symbols = np.asarray(symbols) * np.sqrt(10)

    output_bits = []

    levels = np.array([-3, -1, 1, 3])

    for s in symbols:

        i_level = levels[np.argmin(np.abs(levels - s.real))]
        q_level = levels[np.argmin(np.abs(levels - s.imag))]

        output_bits.extend(level_to_gray(i_level))
        output_bits.extend(level_to_gray(q_level))

    return np.array(output_bits)


# ============================================================
# 5. OFDM MODULATOR
# ============================================================

def ofdm_modulate(symbols):

    # Arrange symbols into OFDM blocks
    n_blocks = len(symbols) // N_SUBCARRIERS

    symbols = symbols[:n_blocks * N_SUBCARRIERS]

    grid = symbols.reshape(n_blocks, N_SUBCARRIERS)

    waveform = []

    for block in grid:

        time_domain = np.fft.ifft(block)

        cyclic_prefix = time_domain[-CP_LEN:]

        ofdm_symbol = np.concatenate(
            [cyclic_prefix, time_domain]
        )

        waveform.extend(ofdm_symbol)

    return np.asarray(waveform)


# ============================================================
# 6. OFDM DEMODULATOR
# ============================================================

def ofdm_demodulate(waveform):

    symbol_length = N_SUBCARRIERS + CP_LEN

    n_blocks = len(waveform) // symbol_length

    waveform = waveform[:n_blocks * symbol_length]

    output = []

    for k in range(n_blocks):

        block = waveform[
            k * symbol_length:
            (k + 1) * symbol_length
        ]

        block = block[CP_LEN:]

        freq_domain = np.fft.fft(block)

        output.extend(freq_domain)

    return np.asarray(output)


# ============================================================
# 7. 3GPP-INSPIRED TDL MULTIPATH CHANNEL
#
# NOTE:
# This is a research prototype.
# Replace this function with your exact 3GPP implementation
# when you provide it.
# ============================================================

def generate_3gpp_inspired_channel(
        n_samples,
        num_paths=NUM_PATHS,
        max_delay=MAX_DELAY_SAMPLES,
        doppler_hz=100,
        los_probability=0.8):

    delays = np.sort(
        rng.integers(
            0,
            max_delay + 1,
            size=num_paths
        )
    )

    # Exponential power delay profile
    powers = np.exp(-delays / 3.0)
    powers /= np.sum(powers)

    # Random complex path coefficients
    path_coefficients = (
        rng.normal(size=num_paths)
        + 1j * rng.normal(size=num_paths)
    )

    path_coefficients *= np.sqrt(powers / 2)

    # Rician LoS component
    if rng.random() < los_probability:

        K_factor = 10 ** (8 / 10)

        los = np.sqrt(K_factor / (K_factor + 1))

        nlos_scale = np.sqrt(1 / (K_factor + 1))

        path_coefficients = (
            los
            + nlos_scale * path_coefficients
        )

    # Time-varying Doppler phase
    t = np.arange(n_samples) / FS

    channel = np.zeros(n_samples, dtype=complex)

    for delay, coeff in zip(
            delays,
            path_coefficients):

        doppler_phase = np.exp(
            1j * 2 * np.pi * doppler_hz * t
        )

        if delay == 0:

            delayed_signal = doppler_phase

        else:

            delayed_signal = np.concatenate(
                [
                    np.zeros(delay),
                    doppler_phase[:-delay]
                ]
            )

        channel += coeff * delayed_signal

    channel /= np.sqrt(
        np.mean(np.abs(channel) ** 2) + 1e-12
    )

    return channel, delays, path_coefficients


# ============================================================
# 8. MULTIPATH + DOPPLER CHANNEL
# ============================================================

def apply_multipath_doppler(
        signal,
        doppler_hz,
        los_probability=0.8):

    n = len(signal)

    channel, delays, coefficients = \
        generate_3gpp_inspired_channel(
            n_samples=n,
            doppler_hz=doppler_hz,
            los_probability=los_probability
        )

    output = np.zeros_like(signal, dtype=complex)

    for delay, coeff in zip(
            delays,
            coefficients):

        if delay == 0:

            delayed = signal

        else:

            delayed = np.concatenate(
                [
                    np.zeros(delay, dtype=complex),
                    signal[:-delay]
                ]
            )

        t = np.arange(n) / FS

        doppler_rotation = np.exp(
            1j * 2 * np.pi * doppler_hz * t
        )

        output += (
            coeff
            * delayed
            * doppler_rotation
        )

    return output, channel


# ============================================================
# 9. CARRIER FREQUENCY OFFSET
# ============================================================

def apply_cfo(signal, cfo_normalized):

    n = np.arange(len(signal))

    rotation = np.exp(
        1j * 2 * np.pi *
        cfo_normalized *
        n
    )

    return signal * rotation


# ============================================================
# 10. POWER AMPLIFIER NONLINEARITY
#
# Rapp-like soft clipping model
# ============================================================

def apply_pa_nonlinearity(
        signal,
        alpha=PA_NONLINEARITY):

    magnitude = np.abs(signal)

    phase = np.angle(signal)

    normalized_amplitude = magnitude / (
        np.sqrt(
            np.mean(magnitude ** 2)
        ) + 1e-12
    )

    compressed_amplitude = (
        normalized_amplitude /
        np.sqrt(
            1 +
            alpha *
            normalized_amplitude ** 4
        )
    )

    output = (
        compressed_amplitude *
        np.exp(1j * phase)
    )

    return output


# ============================================================
# 11. MIXER IQ IMBALANCE
# ============================================================

def apply_mixer_iq_imbalance(
        signal,
        gain_imbalance_db=
        IQ_GAIN_IMBALANCE_DB,
        phase_imbalance_deg=
        IQ_PHASE_IMBALANCE_DEG):

    gain_ratio = 10 ** (
        gain_imbalance_db / 20
    )

    phase = np.deg2rad(
        phase_imbalance_deg
    )

    i = signal.real
    q = signal.imag

    i_new = gain_ratio * i

    q_new = (
        np.sin(phase) * i
        + np.cos(phase) * q
    )

    return i_new + 1j * q_new


# ============================================================
# 12. LNA MODEL
# ============================================================

def apply_lna(
        signal,
        gain=1.0,
        noise_factor=0.02,
        nonlinear_strength=LNA_NONLINEARITY):

    amplified = gain * signal

    # LNA nonlinear distortion
    magnitude = np.abs(amplified)

    distortion = (
        nonlinear_strength *
        magnitude ** 2 *
        amplified
    )

    amplified = amplified - distortion

    # LNA internal noise
    noise = np.sqrt(
        noise_factor / 2
    ) * (
        rng.normal(size=len(signal))
        +
        1j * rng.normal(size=len(signal))
    )

    return amplified + noise


# ============================================================
# 13. AWGN
# ============================================================

def add_awgn(signal, snr_db):

    signal_power = np.mean(
        np.abs(signal) ** 2
    )

    snr_linear = 10 ** (
        snr_db / 10
    )

    noise_power = (
        signal_power /
        snr_linear
    )

    noise = np.sqrt(
        noise_power / 2
    ) * (
        rng.normal(size=len(signal))
        +
        1j * rng.normal(size=len(signal))
    )

    return signal + noise


# ============================================================
# 14. COLORED NOISE
#
# AR(1) low-frequency correlated noise
# ============================================================

def generate_colored_noise(
        n_samples,
        strength=COLORED_NOISE_STRENGTH,
        correlation=0.9):

    white = (
        rng.normal(size=n_samples)
        +
        1j * rng.normal(size=n_samples)
    )

    b = [1]

    a = [1, -correlation]

    colored = lfilter(
        b,
        a,
        white
    )

    colored /= np.sqrt(
        np.mean(
            np.abs(colored) ** 2
        ) + 1e-12
    )

    colored *= strength

    return colored


def add_colored_noise(
        signal,
        strength=COLORED_NOISE_STRENGTH):

    noise = generate_colored_noise(
        len(signal),
        strength
    )

    return signal + noise


# ============================================================
# 15. COMPLETE CHANNEL
# ============================================================

def wireless_channel(
        tx_signal,
        snr_db,
        doppler_hz,
        cfo,
        colored_noise_strength,
        los_probability=0.8):

    # --------------------------------
    # 1. Multipath + Doppler
    # --------------------------------

    signal, channel = \
        apply_multipath_doppler(
            tx_signal,
            doppler_hz,
            los_probability
        )

    # --------------------------------
    # 2. CFO
    # --------------------------------

    signal = apply_cfo(
        signal,
        cfo
    )

    # --------------------------------
    # 3. PA nonlinear distortion
    # --------------------------------

    signal = apply_pa_nonlinearity(
        signal
    )

    # --------------------------------
    # 4. Mixer impairment
    # --------------------------------

    signal = apply_mixer_iq_imbalance(
        signal
    )

    # --------------------------------
    # 5. LNA
    # --------------------------------

    signal = apply_lna(
        signal
    )

    # --------------------------------
    # 6. AWGN
    # --------------------------------

    signal = add_awgn(
        signal,
        snr_db
    )

    # --------------------------------
    # 7. Colored noise
    # --------------------------------

    signal = add_colored_noise(
        signal,
        colored_noise_strength
    )

    return signal, channel


# ============================================================
# 16. BUILD EQUALIZER INPUT FEATURES
#
# The MLP receives neighboring I/Q samples.
# ============================================================

def create_equalizer_features(
        received_symbols,
        memory=EQUALIZER_MEMORY):

    x = []

    for i in range(
            memory,
            len(received_symbols) - memory):

        window = received_symbols[
            i - memory:
            i + memory + 1
        ]

        features = []

        for value in window:

            features.append(value.real)
            features.append(value.imag)

        x.append(features)

    return np.asarray(x)


# ============================================================
# 17. CREATE TRAINING TARGETS
# ============================================================

def create_equalizer_targets(
        transmitted_symbols,
        memory=EQUALIZER_MEMORY):

    y = []

    for i in range(
            memory,
            len(transmitted_symbols) - memory):

        symbol = transmitted_symbols[i]

        y.append([
            symbol.real,
            symbol.imag
        ])

    return np.asarray(y)


# ============================================================
# 18. GENERATE ONE TRAINING SAMPLE
# ============================================================

def generate_training_example():

    # Random number of OFDM symbols
    n_ofdm = rng.integers(
        5,
        10
    )

    n_symbols = (
        n_ofdm *
        N_SUBCARRIERS
    )

    n_bits = (
        n_symbols *
        BITS_PER_SYMBOL
    )

    bits = rng.integers(
        0,
        2,
        n_bits
    )

    tx_symbols = bits_to_16qam(bits)

    tx_waveform = ofdm_modulate(
        tx_symbols
    )

    # Random channel conditions
    snr_db = rng.uniform(
        TRAIN_SNR_DB[0],
        TRAIN_SNR_DB[1]
    )

    doppler = rng.uniform(
        DOPPLER_MIN_HZ,
        DOPPLER_MAX_HZ
    )

    cfo = rng.uniform(
        CFO_MIN,
        CFO_MAX
    )

    colored_strength = rng.uniform(
        0.05,
        COLORED_NOISE_STRENGTH
    )

    rx_waveform, channel = \
        wireless_channel(
            tx_waveform,
            snr_db,
            doppler,
            cfo,
            colored_strength
        )

    # OFDM demodulation
    rx_symbols = ofdm_demodulate(
        rx_waveform
    )

    # Match lengths
    n = min(
        len(rx_symbols),
        len(tx_symbols)
    )

    rx_symbols = rx_symbols[:n]
    tx_symbols = tx_symbols[:n]

    X = create_equalizer_features(
        rx_symbols
    )

    Y = create_equalizer_targets(
        tx_symbols
    )

    return X, Y


# ============================================================
# 19. GENERATE DATASET
# ============================================================

def generate_dataset(
        n_examples):

    X_all = []
    Y_all = []

    print(
        f"Generating {n_examples} training examples..."
    )

    for i in range(n_examples):

        X, Y = generate_training_example()

        X_all.append(X)
        Y_all.append(Y)

        if (i + 1) % 25 == 0:

            print(
                f"Generated {i + 1}/{n_examples}"
            )

    X_all = np.vstack(X_all)
    Y_all = np.vstack(Y_all)

    return X_all, Y_all


# ============================================================
# 20. BUILD MLP EQUALIZER
# ============================================================

def build_mlp_equalizer():

    model = Pipeline(
        [
            (
                "scaler",
                StandardScaler()
            ),

            (
                "mlp",
                MLPRegressor(
                    hidden_layer_sizes=(
                        128,
                        64,
                        32
                    ),

                    activation="relu",

                    solver="adam",

                    learning_rate_init=0.001,

                    batch_size=256,

                    max_iter=40,

                    early_stopping=True,

                    validation_fraction=0.10,

                    random_state=SEED,

                    verbose=True
                )
            )
        ]
    )

    return model


# ============================================================
# 21. TRAIN MLP
# ============================================================

print("\n==============================")
print("GENERATING TRAINING DATA")
print("==============================\n")

X_train, Y_train = generate_dataset(
    N_TRAIN_OFDM
)

print("\nTraining dataset shape:")
print("X:", X_train.shape)
print("Y:", Y_train.shape)


print("\n==============================")
print("TRAINING MLP EQUALIZER")
print("==============================\n")

mlp_equalizer = build_mlp_equalizer()

mlp_equalizer.fit(
    X_train,
    Y_train
)

print("\nMLP training completed.")


# ============================================================
# 22. TEST SYSTEM
# ============================================================

def test_system(
        snr_db=TEST_SNR_DB,
        doppler_hz=100,
        cfo=0.005,
        colored_noise_strength=0.2):

    n_symbols = (
        N_TEST_OFDM *
        N_SUBCARRIERS
    )

    n_bits = (
        n_symbols *
        BITS_PER_SYMBOL
    )

    bits = rng.integers(
        0,
        2,
        n_bits
    )

    tx_symbols = bits_to_16qam(bits)

    tx_waveform = ofdm_modulate(
        tx_symbols
    )

    rx_waveform, channel = \
        wireless_channel(
            tx_waveform,
            snr_db,
            doppler_hz,
            cfo,
            colored_noise_strength
        )

    rx_symbols = ofdm_demodulate(
        rx_waveform
    )

    n = min(
        len(tx_symbols),
        len(rx_symbols)
    )

    tx_symbols = tx_symbols[:n]
    rx_symbols = rx_symbols[:n]

    # --------------------------------
    # Baseline receiver
    # --------------------------------

    baseline_bits = demodulate_16qam(
        rx_symbols
    )

    baseline_bits = baseline_bits[
        :len(bits)
    ]

    original_bits = bits[
        :len(baseline_bits)
    ]

    baseline_ber = np.mean(
        baseline_bits != original_bits
    )

    # --------------------------------
    # MLP equalizer
    # --------------------------------

    X_test = create_equalizer_features(
        rx_symbols
    )

    estimated = mlp_equalizer.predict(
        X_test
    )

    equalized_symbols = (
        estimated[:, 0]
        +
        1j * estimated[:, 1]
    )

    equalized_bits = demodulate_16qam(
        equalized_symbols
    )

    target_bits = demodulate_16qam(
        tx_symbols[
            EQUALIZER_MEMORY:
            len(tx_symbols) -
            EQUALIZER_MEMORY
        ]
    )

    nbits = min(
        len(equalized_bits),
        len(target_bits)
    )

    equalized_ber = np.mean(
        equalized_bits[:nbits]
        !=
        target_bits[:nbits]
    )

    # --------------------------------
    # EVM
    # --------------------------------

    reference = tx_symbols[
        EQUALIZER_MEMORY:
        EQUALIZER_MEMORY +
        len(equalized_symbols)
    ]

    n_evm = min(
        len(reference),
        len(equalized_symbols)
    )

    error = (
        equalized_symbols[:n_evm]
        -
        reference[:n_evm]
    )

    evm = (
        np.sqrt(
            np.mean(
                np.abs(error) ** 2
            )
        )
        /
        np.sqrt(
            np.mean(
                np.abs(
                    reference[:n_evm]
                ) ** 2
            )
        )
    )

    return {
        "baseline_ber": baseline_ber,
        "mlp_ber": equalized_ber,
        "evm": evm,
        "tx_symbols": tx_symbols,
        "rx_symbols": rx_symbols,
        "equalized_symbols":
            equalized_symbols
    }


# ============================================================
# 23. SINGLE TEST
# ============================================================

print("\n==============================")
print("TESTING MLP EQUALIZER")
print("==============================\n")

result = test_system()

print(
    f"Baseline BER : "
    f"{result['baseline_ber']:.6f}"
)

print(
    f"MLP BER      : "
    f"{result['mlp_ber']:.6f}"
)

print(
    f"MLP EVM      : "
    f"{result['evm']:.6f}"
)


# ============================================================
# 24. BER VS SNR
# ============================================================

snr_values = np.arange(
    5,
    26,
    2
)

baseline_ber_results = []
mlp_ber_results = []

print("\n==============================")
print("BER VS SNR")
print("==============================\n")

for snr in snr_values:

    result = test_system(
        snr_db=snr,
        doppler_hz=100,
        cfo=0.005,
        colored_noise_strength=0.2
    )

    baseline_ber_results.append(
        result["baseline_ber"]
    )

    mlp_ber_results.append(
        result["mlp_ber"]
    )

    print(
        f"SNR={snr:2d} dB | "
        f"Baseline BER="
        f"{result['baseline_ber']:.5f} | "
        f"MLP BER="
        f"{result['mlp_ber']:.5f}"
    )


# ============================================================
# 25. PLOT BER VS SNR
# ============================================================

plt.figure(figsize=(8, 5))

plt.semilogy(
    snr_values,
    baseline_ber_results,
    "o-",
    label="Unprocessed Receiver"
)

plt.semilogy(
    snr_values,
    mlp_ber_results,
    "s-",
    label="MLP Equalizer"
)

plt.xlabel("SNR (dB)")
plt.ylabel("BER")

plt.title(
    "BER Performance of MLP Equalizer"
)

plt.grid(True, which="both")

plt.legend()

plt.tight_layout()

plt.show()


# ============================================================
# 26. BER VS DOPPLER
# ============================================================

doppler_values = [
    0,
    50,
    100,
    150,
    200,
    250,
    300
]

doppler_baseline = []
doppler_mlp = []

print("\n==============================")
print("BER VS DOPPLER")
print("==============================\n")

for doppler in doppler_values:

    result = test_system(
        snr_db=15,
        doppler_hz=doppler,
        cfo=0.005,
        colored_noise_strength=0.2
    )

    doppler_baseline.append(
        result["baseline_ber"]
    )

    doppler_mlp.append(
        result["mlp_ber"]
    )

    print(
        f"Doppler={doppler:3d} Hz | "
        f"Baseline="
        f"{result['baseline_ber']:.5f} | "
        f"MLP="
        f"{result['mlp_ber']:.5f}"
    )


# ============================================================
# 27. PLOT BER VS DOPPLER
# ============================================================

plt.figure(figsize=(8, 5))

plt.semilogy(
    doppler_values,
    doppler_baseline,
    "o-",
    label="Unprocessed Receiver"
)

plt.semilogy(
    doppler_values,
    doppler_mlp,
    "s-",
    label="MLP Equalizer"
)

plt.xlabel("Doppler Frequency (Hz)")
plt.ylabel("BER")

plt.title(
    "BER vs Doppler Frequency"
)

plt.grid(True, which="both")

plt.legend()

plt.tight_layout()

plt.show()


# ============================================================
# 28. BER VS CFO
# ============================================================

cfo_values = np.array([
    0,
    0.002,
    0.005,
    0.008,
    0.01,
    0.015,
    0.02
])

cfo_baseline = []
cfo_mlp = []

print("\n==============================")
print("BER VS CFO")
print("==============================\n")

for cfo in cfo_values:

    result = test_system(
        snr_db=15,
        doppler_hz=100,
        cfo=cfo,
        colored_noise_strength=0.2
    )

    cfo_baseline.append(
        result["baseline_ber"]
    )

    cfo_mlp.append(
        result["mlp_ber"]
    )

    print(
        f"CFO={cfo:.3f} | "
        f"Baseline="
        f"{result['baseline_ber']:.5f} | "
        f"MLP="
        f"{result['mlp_ber']:.5f}"
    )


# ============================================================
# 29. PLOT BER VS CFO
# ============================================================

plt.figure(figsize=(8, 5))

plt.semilogy(
    cfo_values,
    cfo_baseline,
    "o-",
    label="Unprocessed Receiver"
)

plt.semilogy(
    cfo_values,
    cfo_mlp,
    "s-",
    label="MLP Equalizer"
)

plt.xlabel("Normalized CFO")

plt.ylabel("BER")

plt.title(
    "BER vs Carrier Frequency Offset"
)

plt.grid(True, which="both")

plt.legend()

plt.tight_layout()

plt.show()


# ============================================================
# 30. BER VS COLORED NOISE
# ============================================================

colored_values = np.array([
    0.0,
    0.05,
    0.10,
    0.15,
    0.20,
    0.30,
    0.40
])

colored_baseline = []
colored_mlp = []

print("\n==============================")
print("BER VS COLORED NOISE")
print("==============================\n")

for strength in colored_values:

    result = test_system(
        snr_db=15,
        doppler_hz=100,
        cfo=0.005,
        colored_noise_strength=strength
    )

    colored_baseline.append(
        result["baseline_ber"]
    )

    colored_mlp.append(
        result["mlp_ber"]
    )

    print(
        f"Colored noise={strength:.2f} | "
        f"Baseline="
        f"{result['baseline_ber']:.5f} | "
        f"MLP="
        f"{result['mlp_ber']:.5f}"
    )


# ============================================================
# 31. PLOT BER VS COLORED NOISE
# ============================================================

plt.figure(figsize=(8, 5))

plt.semilogy(
    colored_values,
    colored_baseline,
    "o-",
    label="Unprocessed Receiver"
)

plt.semilogy(
    colored_values,
    colored_mlp,
    "s-",
    label="MLP Equalizer"
)

plt.xlabel(
    "Colored Noise Strength"
)

plt.ylabel("BER")

plt.title(
    "BER vs Colored Noise"
)

plt.grid(True, which="both")

plt.legend()

plt.tight_layout()

plt.show()


# ============================================================
# 32. CONSTELLATION BEFORE / AFTER MLP
# ============================================================

plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)

plt.scatter(
    result["rx_symbols"].real[:2000],
    result["rx_symbols"].imag[:2000],
    s=5
)

plt.xlabel("In-phase")
plt.ylabel("Quadrature")

plt.title(
    "Received Constellation"
)

plt.grid(True)


plt.subplot(1, 2, 2)

plt.scatter(
    result["equalized_symbols"].real[:2000],
    result["equalized_symbols"].imag[:2000],
    s=5
)

plt.xlabel("In-phase")
plt.ylabel("Quadrature")

plt.title(
    "After MLP Equalization"
)

plt.grid(True)

plt.tight_layout()

plt.show()


# ============================================================
# 33. FINAL SUMMARY
# ============================================================

print("\n")
print("=" * 60)
print("FINAL SYSTEM SUMMARY")
print("=" * 60)

print("""
Modulation:
    16-QAM

Waveform:
    OFDM

Channel:
    Multipath + Doppler
    3GPP-inspired TDL prototype

Noise:
    AWGN
    Colored noise

RF impairments:
    Carrier Frequency Offset
    PA nonlinear distortion
    Mixer IQ imbalance
    LNA noise
    LNA nonlinear distortion

Receiver:
    MLP Equalizer

MLP architecture:
    128 -> 64 -> 32 -> 2

Main metrics:
    BER
    EVM
    Constellation recovery

Experiments:
    BER vs SNR
    BER vs Doppler
    BER vs CFO
    BER vs colored noise
""")

print("=" * 60)
```

```python
# ============================================================
# DBSCAN + UAV DEPLOYMENT + 16-QAM + OFDM + CHANNEL
# + RF IMPAIRMENTS + AWGN/COLORED NOISE + MLP EQUALIZER
#
# Execution order:
#
# 1. Generate emergency users
# 2. DBSCAN clustering
# 3. Identify outliers
# 4. Deploy cluster UAVs
# 5. Place CBS
# 6. Generate 16-QAM symbols
# 7. OFDM modulation
# 8. Multipath + Doppler channel
# 9. CFO
# 10. PA nonlinearity
# 11. Mixer IQ imbalance
# 12. LNA impairment
# 13. AWGN + colored noise
# 14. OFDM demodulation
# 15. MLP equalizer
# 16. 16-QAM demodulation
# 17. BER / EVM
# 18. BER vs SNR / Doppler / CFO / colored noise
#
# ============================================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from scipy.signal import lfilter
from sklearn.cluster import DBSCAN
from sklearn.neural_network import MLPRegressor
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline


# ============================================================
# 1. RANDOM SEED
# ============================================================

SEED = 42
rng = np.random.default_rng(SEED)


# ============================================================
# 2. SYSTEM PARAMETERS
# ============================================================

# -----------------------------
# Disaster area
# -----------------------------

AREA_X = 1000.0       # meters
AREA_Y = 1000.0       # meters

NUM_USERS = 500

# -----------------------------
# DBSCAN
# -----------------------------

DBSCAN_EPS = 90.0
DBSCAN_MIN_SAMPLES = 12

# -----------------------------
# UAV
# -----------------------------

UAV_HEIGHT = 150.0    # meters

# -----------------------------
# 16-QAM
# -----------------------------

BITS_PER_SYMBOL = 4

# -----------------------------
# OFDM
# -----------------------------

N_SUBCARRIERS = 64
CP_LEN = 16

# -----------------------------
# MLP
# -----------------------------

MEMORY = 2

# -----------------------------
# Sampling
# -----------------------------

FS = 1e6


# ============================================================
# 3. GENERATE EMERGENCY USERS
# ============================================================

def generate_users():

    """
    Generates a mixture of dense disaster-area user clusters
    and isolated/scattered users.

    This is important because DBSCAN should be able to identify
    the isolated users as noise/outliers.
    """

    centers = np.array([
        [180, 180],
        [500, 200],
        [800, 250],
        [250, 600],
        [650, 650],
        [850, 800]
    ])

    users_per_cluster = 70

    all_users = []

    for center in centers:

        cluster = center + rng.normal(
            0,
            45,
            size=(users_per_cluster, 2)
        )

        all_users.append(cluster)

    users = np.vstack(all_users)

    # Add scattered users
    n_scattered = NUM_USERS - len(users)

    scattered = rng.uniform(
        low=[0, 0],
        high=[AREA_X, AREA_Y],
        size=(n_scattered, 2)
    )

    users = np.vstack([
        users,
        scattered
    ])

    # Keep inside area
    users[:, 0] = np.clip(
        users[:, 0],
        0,
        AREA_X
    )

    users[:, 1] = np.clip(
        users[:, 1],
        0,
        AREA_Y
    )

    return users


# ============================================================
# 4. DBSCAN CLUSTERING
# ============================================================

def perform_dbscan(users):

    model = DBSCAN(
        eps=DBSCAN_EPS,
        min_samples=DBSCAN_MIN_SAMPLES
    )

    labels = model.fit_predict(users)

    return labels


# ============================================================
# 5. EXTRACT CLUSTERS
# ============================================================

def extract_clusters(users, labels):

    cluster_ids = sorted(
        [x for x in np.unique(labels) if x != -1]
    )

    clusters = []

    for cluster_id in cluster_ids:

        cluster_users = users[
            labels == cluster_id
        ]

        clusters.append(
            cluster_users
        )

    return clusters


# ============================================================
# 6. FIND OUTLIERS
# ============================================================

def find_outliers(users, labels):

    return users[
        labels == -1
    ]


# ============================================================
# 7. CALCULATE UAV LOCATIONS
# ============================================================

def calculate_uav_positions(clusters):

    positions = []

    for cluster in clusters:

        center = np.mean(
            cluster,
            axis=0
        )

        positions.append(center)

    return np.array(positions)


# ============================================================
# 8. CALCULATE CBS POSITION
# ============================================================

def calculate_cbs_position(uav_positions):

    return np.mean(
        uav_positions,
        axis=0
    )


# ============================================================
# 9. PLOT DBSCAN DEPLOYMENT
# ============================================================

def plot_deployment(
        users,
        labels,
        uav_positions,
        cbs_position):

    plt.figure(figsize=(9, 7))

    unique_labels = np.unique(labels)

    for label in unique_labels:

        if label == -1:

            points = users[
                labels == label
            ]

            plt.scatter(
                points[:, 0],
                points[:, 1],
                marker="x",
                s=40,
                label="DBSCAN Outliers"
            )

        else:

            points = users[
                labels == label
            ]

            plt.scatter(
                points[:, 0],
                points[:, 1],
                s=15,
                label=f"Cluster {label}"
            )

    plt.scatter(
        uav_positions[:, 0],
        uav_positions[:, 1],
        marker="^",
        s=150,
        edgecolors="black",
        label="Cluster UAV"
    )

    plt.scatter(
        cbs_position[0],
        cbs_position[1],
        marker="*",
        s=250,
        edgecolors="black",
        label="CBS"
    )

    plt.xlabel("X position (m)")
    plt.ylabel("Y position (m)")

    plt.title(
        "DBSCAN-Based UAV Deployment"
    )

    plt.xlim(0, AREA_X)
    plt.ylim(0, AREA_Y)

    plt.grid(True)
    plt.legend(
        bbox_to_anchor=(1.04, 1),
        loc="upper left"
    )

    plt.tight_layout()
    plt.show()


# ============================================================
# 10. 16-QAM MAPPING
# ============================================================

def bits_to_16qam(bits):

    bits = np.asarray(bits)

    bits = bits.reshape(
        -1,
        4
    )

    mapping = {
        (0, 0): -3,
        (0, 1): -1,
        (1, 1): 1,
        (1, 0): 3
    }

    symbols = []

    for b in bits:

        i = mapping[
            (b[0], b[1])
        ]

        q = mapping[
            (b[2], b[3])
        ]

        symbols.append(
            i + 1j * q
        )

    symbols = np.array(
        symbols,
        dtype=complex
    )

    # Normalize average symbol power
    symbols /= np.sqrt(10)

    return symbols


# ============================================================
# 11. 16-QAM DEMAPPING
# ============================================================

def demodulate_16qam(symbols):

    symbols = (
        symbols *
        np.sqrt(10)
    )

    levels = np.array([
        -3,
        -1,
        1,
        3
    ])

    output = []

    for symbol in symbols:

        i = levels[
            np.argmin(
                np.abs(
                    levels -
                    symbol.real
                )
            )
        ]

        q = levels[
            np.argmin(
                np.abs(
                    levels -
                    symbol.imag
                )
            )
        ]

        if i == -3:
            output.extend([0, 0])
        elif i == -1:
            output.extend([0, 1])
        elif i == 1:
            output.extend([1, 1])
        else:
            output.extend([1, 0])

        if q == -3:
            output.extend([0, 0])
        elif q == -1:
            output.extend([0, 1])
        elif q == 1:
            output.extend([1, 1])
        else:
            output.extend([1, 0])

    return np.array(
        output,
        dtype=int
    )


# ============================================================
# 12. OFDM MODULATION
# ============================================================

def ofdm_modulate(symbols):

    n_blocks = (
        len(symbols)
        // N_SUBCARRIERS
    )

    symbols = symbols[
        :n_blocks *
        N_SUBCARRIERS
    ]

    grid = symbols.reshape(
        n_blocks,
        N_SUBCARRIERS
    )

    output = []

    for block in grid:

        time_signal = np.fft.ifft(
            block
        )

        cp = time_signal[
            -CP_LEN:
        ]

        ofdm_symbol = np.concatenate(
            [cp, time_signal]
        )

        output.extend(
            ofdm_symbol
        )

    return np.array(
        output,
        dtype=complex
    )


# ============================================================
# 13. OFDM DEMODULATION
# ============================================================

def ofdm_demodulate(signal):

    block_length = (
        N_SUBCARRIERS +
        CP_LEN
    )

    n_blocks = (
        len(signal)
        // block_length
    )

    signal = signal[
        :n_blocks *
        block_length
    ]

    output = []

    for k in range(n_blocks):

        start = (
            k *
            block_length
        )

        end = (
            (k + 1) *
            block_length
        )

        block = signal[
            start:end
        ]

        block = block[
            CP_LEN:
        ]

        frequency_signal = np.fft.fft(
            block
        )

        output.extend(
            frequency_signal
        )

    return np.array(
        output,
        dtype=complex
    )


# ============================================================
# 14. MULTIPATH CHANNEL
# ============================================================

def multipath_channel(
        signal,
        num_paths=5):

    delays = np.array([
        0,
        1,
        2,
        4,
        7
    ])

    powers = np.array([
        0.40,
        0.25,
        0.15,
        0.12,
        0.08
    ])

    powers = (
        powers /
        np.sum(powers)
    )

    output = np.zeros(
        len(signal),
        dtype=complex
    )

    for delay, power in zip(
            delays,
            powers):

        h = (
            rng.normal()
            +
            1j * rng.normal()
        ) / np.sqrt(2)

        h *= np.sqrt(power)

        if delay == 0:

            delayed = signal

        else:

            delayed = np.concatenate([
                np.zeros(
                    delay,
                    dtype=complex
                ),
                signal[:-delay]
            ])

        output += (
            h *
            delayed
        )

    return output


# ============================================================
# 15. DOPPLER
# ============================================================

def apply_doppler(
        signal,
        doppler_hz):

    n = np.arange(
        len(signal)
    )

    t = n / FS

    phase = np.exp(
        1j *
        2 *
        np.pi *
        doppler_hz *
        t
    )

    return signal * phase


# ============================================================
# 16. CFO
# ============================================================

def apply_cfo(
        signal,
        normalized_cfo):

    n = np.arange(
        len(signal)
    )

    phase = np.exp(
        1j *
        2 *
        np.pi *
        normalized_cfo *
        n
    )

    return signal * phase


# ============================================================
# 17. PA NONLINEARITY
# ============================================================

def apply_pa_nonlinearity(
        signal,
        alpha=0.15):

    power = np.abs(signal) ** 2

    return signal * (
        1 -
        alpha *
        power
    )


# ============================================================
# 18. MIXER IQ IMBALANCE
# ============================================================

def apply_mixer_impairment(
        signal,
        gain_imbalance_db=0.5,
        phase_error_deg=3):

    gain_ratio = 10 ** (
        gain_imbalance_db /
        20
    )

    phase = np.deg2rad(
        phase_error_deg
    )

    i = signal.real
    q = signal.imag

    i_new = (
        gain_ratio *
        i
    )

    q_new = (
        q *
        np.cos(phase)
        +
        i *
        np.sin(phase)
    )

    return (
        i_new +
        1j * q_new
    )


# ============================================================
# 19. LNA IMPAIRMENT
# ============================================================

def apply_lna(
        signal,
        gain=1.0,
        internal_noise=0.01):

    amplified = (
        gain *
        signal
    )

    noise = np.sqrt(
        internal_noise / 2
    ) * (
        rng.normal(
            size=len(signal)
        )
        +
        1j *
        rng.normal(
            size=len(signal)
        )
    )

    # Mild LNA nonlinearity
    nonlinear = (
        0.02 *
        np.abs(amplified) ** 2 *
        amplified
    )

    return (
        amplified
        -
        nonlinear
        +
        noise
    )


# ============================================================
# 20. AWGN
# ============================================================

def add_awgn(
        signal,
        snr_db):

    signal_power = np.mean(
        np.abs(signal) ** 2
    )

    snr_linear = (
        10 **
        (snr_db / 10)
    )

    noise_power = (
        signal_power /
        snr_linear
    )

    noise = np.sqrt(
        noise_power / 2
    ) * (
        rng.normal(
            size=len(signal)
        )
        +
        1j *
        rng.normal(
            size=len(signal)
        )
    )

    return (
        signal +
        noise
    )


# ============================================================
# 21. COLORED NOISE
# ============================================================

def add_colored_noise(
        signal,
        strength=0.05):

    white_noise = (
        rng.normal(
            size=len(signal)
        )
        +
        1j *
        rng.normal(
            size=len(signal)
        )
    )

    # AR(1) filter
    colored = lfilter(
        [1],
        [1, -0.85],
        white_noise
    )

    colored /= (
        np.sqrt(
            np.mean(
                np.abs(colored) ** 2
            )
        )
        + 1e-12
    )

    colored *= strength

    return (
        signal +
        colored
    )


# ============================================================
# 22. COMPLETE CHANNEL
# ============================================================

def complete_channel(
        signal,
        snr_db=15,
        doppler_hz=100,
        cfo=0.002,
        colored_noise=0.05):

    # 1. Multipath
    signal = multipath_channel(
        signal
    )

    # 2. Doppler
    signal = apply_doppler(
        signal,
        doppler_hz
    )

    # 3. CFO
    signal = apply_cfo(
        signal,
        cfo
    )

    # 4. PA
    signal = apply_pa_nonlinearity(
        signal
    )

    # 5. Mixer
    signal = apply_mixer_impairment(
        signal
    )

    # 6. LNA
    signal = apply_lna(
        signal
    )

    # 7. AWGN
    signal = add_awgn(
        signal,
        snr_db
    )

    # 8. Colored noise
    signal = add_colored_noise(
        signal,
        colored_noise
    )

    return signal


# ============================================================
# 23. CREATE MLP FEATURES
# ============================================================

def create_features(
        symbols,
        memory=MEMORY):

    X = []

    for i in range(
            memory,
            len(symbols) - memory):

        window = symbols[
            i-memory:i+memory+1
        ]

        feature = []

        for value in window:

            feature.append(
                value.real
            )

            feature.append(
                value.imag
            )

        X.append(feature)

    return np.array(X)


# ============================================================
# 24. CREATE MLP TARGETS
# ============================================================

def create_targets(
        symbols,
        memory=MEMORY):

    Y = []

    for i in range(
            memory,
            len(symbols) - memory):

        Y.append([
            symbols[i].real,
            symbols[i].imag
        ])

    return np.array(Y)


# ============================================================
# 25. GENERATE TRAINING DATA
# ============================================================

def generate_training_data(
        n_examples=30):

    X_all = []
    Y_all = []

    print(
        "\nGenerating MLP training data..."
    )

    for example in range(
            n_examples):

        n_symbols = (
            N_SUBCARRIERS *
            5
        )

        bits = rng.integers(
            0,
            2,
            n_symbols *
            BITS_PER_SYMBOL
        )

        tx_symbols = bits_to_16qam(
            bits
        )

        tx_signal = ofdm_modulate(
            tx_symbols
        )

        # Random channel conditions
        snr = rng.uniform(
            10,
            25
        )

        doppler = rng.uniform(
            0,
            250
        )

        cfo = rng.uniform(
            -0.01,
            0.01
        )

        rx_signal = complete_channel(
            tx_signal,
            snr_db=snr,
            doppler_hz=doppler,
            cfo=cfo,
            colored_noise=0.05
        )

        rx_symbols = ofdm_demodulate(
            rx_signal
        )

        n = min(
            len(tx_symbols),
            len(rx_symbols)
        )

        tx_symbols = tx_symbols[:n]
        rx_symbols = rx_symbols[:n]

        X = create_features(
            rx_symbols
        )

        Y = create_targets(
            tx_symbols
        )

        X_all.append(X)
        Y_all.append(Y)

        print(
            f"Training sample "
            f"{example + 1}/{n_examples}"
        )

    return (
        np.vstack(X_all),
        np.vstack(Y_all)
    )


# ============================================================
# 26. BUILD MLP
# ============================================================

def build_mlp():

    model = Pipeline([
        (
            "scaler",
            StandardScaler()
        ),

        (
            "mlp",
            MLPRegressor(
                hidden_layer_sizes=(
                    128,
                    64,
                    32
                ),
                activation="relu",
                solver="adam",
                learning_rate_init=0.001,
                batch_size=256,
                max_iter=50,
                early_stopping=True,
                validation_fraction=0.1,
                random_state=SEED,
                verbose=True
            )
        )
    ])

    return model


# ============================================================
# 27. TRAIN MLP
# ============================================================

print("\n")
print("=" * 70)
print("STEP 1: GENERATING EMERGENCY USERS")
print("=" * 70)

users = generate_users()

print(
    "Total users:",
    len(users)
)


print("\n")
print("=" * 70)
print("STEP 2: DBSCAN CLUSTERING")
print("=" * 70)

labels = perform_dbscan(
    users
)

clusters = extract_clusters(
    users,
    labels
)

outliers = find_outliers(
    users,
    labels
)

print(
    "Number of DBSCAN clusters:",
    len(clusters)
)

print(
    "Number of DBSCAN outliers:",
    len(outliers)
)

print(
    "Cluster sizes:",
    [
        len(c)
        for c in clusters
    ]
)


# ============================================================
# 28. UAV DEPLOYMENT
# ============================================================

print("\n")
print("=" * 70)
print("STEP 3: UAV DEPLOYMENT")
print("=" * 70)

uav_positions = calculate_uav_positions(
    clusters
)

cbs_position = calculate_cbs_position(
    uav_positions
)

print("\nCluster UAV positions:")

for i, position in enumerate(
        uav_positions):

    print(
        f"UAV {i + 1}: "
        f"X={position[0]:.2f} m, "
        f"Y={position[1]:.2f} m, "
        f"Height={UAV_HEIGHT:.2f} m"
    )

print(
    "\nCBS position: "
    f"X={cbs_position[0]:.2f} m, "
    f"Y={cbs_position[1]:.2f} m"
)


# ============================================================
# 29. PLOT DBSCAN
# ============================================================

plot_deployment(
    users,
    labels,
    uav_positions,
    cbs_position
)


# ============================================================
# 30. GENERATE TRAINING DATA
# ============================================================

print("\n")
print("=" * 70)
print("STEP 4: TRAINING DATA FOR MLP")
print("=" * 70)

X_train, Y_train = generate_training_data(
    n_examples=30
)

print(
    "\nX training shape:",
    X_train.shape
)

print(
    "Y training shape:",
    Y_train.shape
)


# ============================================================
# 31. TRAIN MLP
# ============================================================

print("\n")
print("=" * 70)
print("STEP 5: TRAINING MLP EQUALIZER")
print("=" * 70)

mlp = build_mlp()

mlp.fit(
    X_train,
    Y_train
)

print(
    "\nMLP training completed."
)


# ============================================================
# 32. TEST FUNCTION
# ============================================================

def run_test(
        snr_db=15,
        doppler_hz=100,
        cfo=0.002,
        colored_noise=0.05):

    n_symbols = (
        N_SUBCARRIERS *
        10
    )

    bits = rng.integers(
        0,
        2,
        n_symbols *
        BITS_PER_SYMBOL
    )

    tx_symbols = bits_to_16qam(
        bits
    )

    tx_signal = ofdm_modulate(
        tx_symbols
    )

    rx_signal = complete_channel(
        tx_signal,
        snr_db,
        doppler_hz,
        cfo,
        colored_noise
    )

    rx_symbols = ofdm_demodulate(
        rx_signal
    )

    n = min(
        len(tx_symbols),
        len(rx_symbols)
    )

    tx_symbols = tx_symbols[:n]
    rx_symbols = rx_symbols[:n]

    # ------------------------------------
    # Baseline receiver
    # ------------------------------------

    baseline_bits = demodulate_16qam(
        rx_symbols
    )

    n_bits = min(
        len(bits),
        len(baseline_bits)
    )

    baseline_ber = np.mean(
        bits[:n_bits]
        !=
        baseline_bits[:n_bits]
    )

    # ------------------------------------
    # MLP equalizer
    # ------------------------------------

    X_test = create_features(
        rx_symbols
    )

    predicted = mlp.predict(
        X_test
    )

    equalized_symbols = (
        predicted[:, 0]
        +
        1j *
        predicted[:, 1]
    )

    equalized_bits = demodulate_16qam(
        equalized_symbols
    )

    target_symbols = tx_symbols[
        MEMORY:
        len(tx_symbols) -
        MEMORY
    ]

    target_bits = demodulate_16qam(
        target_symbols
    )

    n_bits_mlp = min(
        len(equalized_bits),
        len(target_bits)
    )

    mlp_ber = np.mean(
        equalized_bits[:n_bits_mlp]
        !=
        target_bits[:n_bits_mlp]
    )

    # ------------------------------------
    # EVM
    # ------------------------------------

    n_evm = min(
        len(equalized_symbols),
        len(target_symbols)
    )

    error = (
        equalized_symbols[:n_evm]
        -
        target_symbols[:n_evm]
    )

    evm = (
        np.sqrt(
            np.mean(
                np.abs(error) ** 2
            )
        )
        /
        (
            np.sqrt(
                np.mean(
                    np.abs(
                        target_symbols[
                            :n_evm
                        ]
                    ) ** 2
                )
            )
            + 1e-12
        )
    )

    return {
        "baseline_ber": baseline_ber,
        "mlp_ber": mlp_ber,
        "evm": evm,
        "rx_symbols": rx_symbols,
        "equalized_symbols":
            equalized_symbols
    }


# ============================================================
# 33. SINGLE TEST
# ============================================================

print("\n")
print("=" * 70)
print("STEP 6: SINGLE COMMUNICATION TEST")
print("=" * 70)

result = run_test()

print(
    f"\nBaseline BER = "
    f"{result['baseline_ber']:.6f}"
)

print(
    f"MLP BER      = "
    f"{result['mlp_ber']:.6f}"
)

print(
    f"EVM          = "
    f"{result['evm']:.6f}"
)


# ============================================================
# 34. BER VS SNR
# ============================================================

print("\n")
print("=" * 70)
print("STEP 7: BER VS SNR")
print("=" * 70)

snr_values = np.array([
    5,
    8,
    11,
    14,
    17,
    20,
    23
])

baseline_snr = []
mlp_snr = []

for snr in snr_values:

    result = run_test(
        snr_db=snr,
        doppler_hz=100,
        cfo=0.002,
        colored_noise=0.05
    )

    baseline_snr.append(
        result["baseline_ber"]
    )

    mlp_snr.append(
        result["mlp_ber"]
    )

    print(
        f"SNR={snr:2d} dB | "
        f"Baseline={result['baseline_ber']:.6f} | "
        f"MLP={result['mlp_ber']:.6f}"
    )


plt.figure(figsize=(8, 5))

plt.semilogy(
    snr_values,
    baseline_snr,
    "o-",
    label="Without MLP"
)

plt.semilogy(
    snr_values,
    mlp_snr,
    "s-",
    label="With MLP Equalizer"
)

plt.xlabel("SNR (dB)")
plt.ylabel("BER")

plt.title(
    "BER vs SNR"
)

plt.grid(
    True,
    which="both"
)

plt.legend()

plt.tight_layout()

plt.show()


# ============================================================
# 35. BER VS DOPPLER
# ============================================================

print("\n")
print("=" * 70)
print("STEP 8: BER VS DOPPLER")
print("=" * 70)

doppler_values = np.array([
    0,
    50,
    100,
    150,
    200,
    250,
    300
])

baseline_doppler = []
mlp_doppler = []

for doppler in doppler_values:

    result = run_test(
        snr_db=15,
        doppler_hz=doppler,
        cfo=0.002,
        colored_noise=0.05
    )

    baseline_doppler.append(
        result["baseline_ber"]
    )

    mlp_doppler.append(
        result["mlp_ber"]
    )

    print(
        f"Doppler={doppler:3d} Hz | "
        f"Baseline={result['baseline_ber']:.6f} | "
        f"MLP={result['mlp_ber']:.6f}"
    )


plt.figure(figsize=(8, 5))

plt.semilogy(
    doppler_values,
    baseline_doppler,
    "o-",
    label="Without MLP"
)

plt.semilogy(
    doppler_values,
    mlp_doppler,
    "s-",
    label="With MLP Equalizer"
)

plt.xlabel(
    "Doppler Frequency (Hz)"
)

plt.ylabel("BER")

plt.title(
    "BER vs Doppler Frequency"
)

plt.grid(
    True,
    which="both"
)

plt.legend()

plt.tight_layout()

plt.show()


# ============================================================
# 36. BER VS CFO
# ============================================================

print("\n")
print("=" * 70)
print("STEP 9: BER VS CFO")
print("=" * 70)

cfo_values = np.array([
    0,
    0.002,
    0.004,
    0.006,
    0.008,
    0.010
])

baseline_cfo = []
mlp_cfo = []

for cfo in cfo_values:

    result = run_test(
        snr_db=15,
        doppler_hz=100,
        cfo=cfo,
        colored_noise=0.05
    )

    baseline_cfo.append(
        result["baseline_ber"]
    )

    mlp_cfo.append(
        result["mlp_ber"]
    )

    print(
        f"CFO={cfo:.3f} | "
        f"Baseline={result['baseline_ber']:.6f} | "
        f"MLP={result['mlp_ber']:.6f}"
    )


plt.figure(figsize=(8, 5))

plt.semilogy(
    cfo_values,
    baseline_cfo,
    "o-",
    label="Without MLP"
)

plt.semilogy(
    cfo_values,
    mlp_cfo,
    "s-",
    label="With MLP Equalizer"
)

plt.xlabel(
    "Normalized CFO"
)

plt.ylabel("BER")

plt.title(
    "BER vs Carrier Frequency Offset"
)

plt.grid(
    True,
    which="both"
)

plt.legend()

plt.tight_layout()

plt.show()


# ============================================================
# 37. BER VS COLORED NOISE
# ============================================================

print("\n")
print("=" * 70)
print("STEP 10: BER VS COLORED NOISE")
print("=" * 70)

colored_values = np.array([
    0.00,
    0.02,
    0.04,
    0.06,
    0.08,
    0.10
])

baseline_colored = []
mlp_colored = []

for noise in colored_values:

    result = run_test(
        snr_db=15,
        doppler_hz=100,
        cfo=0.002,
        colored_noise=noise
    )

    baseline_colored.append(
        result["baseline_ber"]
    )

    mlp_colored.append(
        result["mlp_ber"]
    )

    print(
        f"Noise={noise:.2f} | "
        f"Baseline={result['baseline_ber']:.6f} | "
        f"MLP={result['mlp_ber']:.6f}"
    )


plt.figure(figsize=(8, 5))

plt.semilogy(
    colored_values,
    baseline_colored,
    "o-",
    label="Without MLP"
)

plt.semilogy(
    colored_values,
    mlp_colored,
    "s-",
    label="With MLP Equalizer"
)

plt.xlabel(
    "Colored Noise Strength"
)

plt.ylabel("BER")

plt.title(
    "BER vs Colored Noise"
)

plt.grid(
    True,
    which="both"
)

plt.legend()

plt.tight_layout()

plt.show()


# ============================================================
# 38. CONSTELLATION COMPARISON
# ============================================================

plt.figure(figsize=(12, 5))

plt.subplot(
    1,
    2,
    1
)

plt.scatter(
    result["rx_symbols"].real[:2000],
    result["rx_symbols"].imag[:2000],
    s=8
)

plt.xlabel("I")
plt.ylabel("Q")

plt.title(
    "Received 16-QAM Constellation"
)

plt.grid(True)


plt.subplot(
    1,
    2,
    2
)

plt.scatter(
    result["equalized_symbols"].real[:2000],
    result["equalized_symbols"].imag[:2000],
    s=8
)

plt.xlabel("I")
plt.ylabel("Q")

plt.title(
    "After MLP Equalization"
)

plt.grid(True)

plt.tight_layout()

plt.show()


# ============================================================
# 39. SAVE DBSCAN RESULTS
# ============================================================

cluster_table = []

for i, cluster in enumerate(
        clusters):

    center = np.mean(
        cluster,
        axis=0
    )

    cluster_table.append({
        "Cluster_ID": i + 1,
        "Users": len(cluster),
        "UAV_X_m": center[0],
        "UAV_Y_m": center[1],
        "UAV_Height_m": UAV_HEIGHT
    })


cluster_df = pd.DataFrame(
    cluster_table
)

cluster_df.to_csv(
    "dbscan_uav_deployment.csv",
    index=False
)


# ============================================================
# 40. SAVE COMMUNICATION RESULTS
# ============================================================

results_df = pd.DataFrame({
    "SNR_dB": snr_values,
    "Baseline_BER": baseline_snr,
    "MLP_BER": mlp_snr
})

results_df.to_csv(
    "mlp_ber_vs_snr.csv",
    index=False
)


print("\n")
print("=" * 70)
print("SIMULATION COMPLETED")
print("=" * 70)

print(
    "\nGenerated files:"
)

print(
    "1. dbscan_uav_deployment.csv"
)

print(
    "2. mlp_ber_vs_snr.csv"
)

print(
    "\nMain outputs:"
)

print(
    "DBSCAN clusters:",
    len(clusters)
)

print(
    "DBSCAN outliers:",
    len(outliers)
)

print(
    "CBS:",
    cbs_position
)

print(
    "\nFinal test:"
)

print(
    "Baseline BER:",
    result["baseline_ber"]
)

print(
    "MLP BER:",
    result["mlp_ber"]
)

print(
    "EVM:",
    result["evm"]
)
```
